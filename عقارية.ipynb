```python
pareto = (
    df.groupby('المشروع')['السعر الإجمالي (جنيه)']
    .sum()
    .sort_values(ascending=False)
)

pareto_df = pd.DataFrame({
    'إجمالي المبيعات': pareto
})

pareto_df['نسبة المساهمة %'] = (
    pareto_df['إجمالي المبيعات'] /
    pareto_df['إجمالي المبيعات'].sum()
) * 100

pareto_df['النسبة التراكمية %'] = (
    pareto_df['نسبة المساهمة %'].cumsum()
)

pareto_df
```

```python
fig, ax1 = plt.subplots(figsize=(12, 7))

# الأعمدة
ax1.bar(
    pareto_df.index,
    pareto_df['إجمالي المبيعات']
)

ax1.set_xlabel('المشروع')
ax1.set_ylabel('إجمالي المبيعات (جنيه)')
ax1.tick_params(axis='x', rotation=45)

# الخط التراكمي
ax2 = ax1.twinx()

ax2.plot(
    pareto_df.index,
    pareto_df['النسبة التراكمية %'],
    marker='o'
)

ax2.set_ylabel('النسبة التراكمية (%)')
ax2.set_ylim(0, 110)

# خط 80%
ax2.axhline(
    80,
    linestyle='--',
    linewidth=1
)

plt.title('تحليل Pareto لمساهمة المشاريع في إجمالي المبيعات')

plt.tight_layout()
plt.show()
```
